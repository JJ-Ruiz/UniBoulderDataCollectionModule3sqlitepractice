# ESTABLISH THE CONNECTION

```python
import sqlite3
connection = sqlite3.connect('/content/bike_rentals.sqlite')

# Creating the cursor object:

cursor = connection.cursor()

# This query is to find the table name:

query = '''
SELECT name
FROM sqlite_master
WHERE type ='table'
'''

cursor.execute(query)
results = cursor.fetchall()
results
```

# PREVIEW THE DATA

```python
# Retrieve only the first 3 rows of the table you found in Task 2 and display them.
# What type of object is the result? What type is each row?

query = '''
SELECT *
FROM bike_rentals_data
limit 3
'''
cursor.execute(query)
results = cursor.fetchall()
results

# What type of object is the result: This objec is a list
type(results)

#What type of object is each row: This, however, is a tuple
type(results[0])
```

# FETCH ALL RECORDS

```python
# Retrieve all rows from the table and store them in a variable called results.
# How many rows are there? (Hint: len()).

query = '''
SELECT *
FROM bike_rentals_data
'''

cursor.execute(query)
results = cursor.fetchall()
results

len(results)
```

# Add the column names

```python
# The rows I fetched contain only values, with no column names. The columns, in
# the same order as the values in each tuple, are:
cols = ['rental_id',
        'user_id',
        'gender',
        'age',
        'bike_type',
        'duration_minutes',
        'fare',
        'payment_method',
        'rental_date',
        'station'
        ]
```

```python
import pandas as pd
df = pd.DataFrame(results, columns =cols)
df.info()
```

# Save the data to a CSV file


```python
# Save the DataFrame to a CSV file called bike_rentals.csv in the
# /content/ folder, so you can use it later. Finally, close the database connection.

df.to_csv('/content/rental_bikes.csv')
connection.close()
```
